### Пояснення поведінки Kernel

Після виконання команди Restart Kernel процес ядра Python скидається до початкового стану. Оперативна пам'ять очищується, через що всі змінні (зокрема змінна `x`), створені під час попередньої сесії, перестають існувати в пам'яті. Щоб змінна знову стала доступною для подальших осередків, осередок з її оголошенням необхідно виконати заново.
